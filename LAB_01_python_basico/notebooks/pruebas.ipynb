```python
import pandas as pd

def pregunta_01():
    """
    
    Calcule la suma de los valores de la segunda columna (`value`) del
    archivo `data/data.csv.gz` y retorne el resultado como un número entero.

    Ejemplo del formato de la respuesta:

        214
    """
    
    data = pd.read_csv("LAB_01_python_basico\data\data.csv.gz", compression='gzip', sep='\t', header=None)
    suma = data[1].sum()   # la segunda columna es el índice 1
    
    return suma

pregunta_01()
```

```python
import pandas as pd

def pregunta_02():
    data = pd.read_csv(
        '../data/data.csv.gz',
        compression='gzip',
        sep='\t',
        header=None,
        names=['letter', 'value', 'date', 'letters', 'pairs'],
    )
    conteo = data['letter'].value_counts().sort_index()
    return list(conteo.items())

pregunta_02()
```
