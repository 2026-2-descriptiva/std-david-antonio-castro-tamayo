```python
# Preparacion

import matplotlib
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Carga de la tabla drivers

drivers = pd.read_csv(
    "../data/drivers.csv",
    sep=",",
    thousands=None,
    decimal=".",
)
drivers.head()
```

```python
# Carga de la tabla timesheet

timesheet = pd.read_csv(
    "../data/timesheet.csv",
    sep=",",
    thousands=None,
    decimal=".",
)
timesheet.head()
```

```python
# Media de las horas y las millas de cada conductor

mean_timesheet = timesheet.groupby("driverId").mean()
mean_timesheet.head()
```
